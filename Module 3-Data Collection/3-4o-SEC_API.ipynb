## Module 3-4o Using SEC EDGAR API

For more information about the `secedgar` package, see: https://sec-edgar.github.io/sec-edgar.

```python
from secedgar import CompanyFilings, FilingType
from secedgar.cik_lookup import CIKLookup
import nest_asyncio
from datetime import date
nest_asyncio.apply()
```

```python
lookups = CIKLookup(['aapl', 'msft', 'facebook'],
                    user_agent="Leonard Li (leye.li@unsw.edu.au)")

lookups.lookup_dict
```

```python
Magnificant7 = {
    "Apple": "0000320193",
    "Microsoft": "0000789019",
    "Amazon": "0001018724",
    "Alphabet": "0001652044",
    "Meta": "0001326801",
    "Nvidia": "0001045810",
    "Tesla": "0001318605"
}
```

```python
filings = CompanyFilings(cik_lookup=list(Magnificant7.values()),
                        filing_type=FilingType.FILING_10K,
                        count=2, 
                        start_date=date(2017, 1, 1),
                        end_date=date(2018, 12, 31), 
                        user_agent='Leonard Li (leye.li@unsw.edu.au)')
```

```python
filings.save('./sec_filings')
```
