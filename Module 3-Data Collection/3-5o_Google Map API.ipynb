### 2. Basic usage of the client

````python
import googlemaps
from dotenv import load_dotenv
load_dotenv()
import os
````

````python
gmaps = googlemaps.Client(key = os.getenv('GOOGLE_MAPS_API'))
````

#### 2.1. Geocoding (address → lat/lng)

````python
geocode_result = gmaps.geocode("255 George St, Sydney NSW 2000, Australia")
print(geocode_result[0]['geometry']['location'])
````

````python
def get_lat_long(address: str) -> tuple:
    geocode_result = gmaps.geocode(address)
    if geocode_result:
        lat = geocode_result[0].get('geometry').get('location').get('lat')
        long = geocode_result[0].get('geometry').get('location').get('lng')
        return lat, long
    return None, None
````

#### 2.2. Reverse Geocoding (lat/lng → address)


````python
reverse_geocode_result = gmaps.reverse_geocode((37.4221, -122.0841))
print(reverse_geocode_result[0]['formatted_address'])
````

### 3. Distance API

#### 3.1. Directions

````python
directions_result = gmaps.directions(
    origin="255 George St, Sydney NSW 2000, Australia",
    destination="159-175 Church St, Parramatta NSW 2150, Australia",
    mode="driving",
    departure_time="now"
)

# Print step-by-step directions
for step in directions_result[0]['legs'][0]['steps']:
    print(step['html_instructions'])
````

#### 3.2. Distance Matrix

````python
matrix = gmaps.distance_matrix(
    origins=["New York, NY"],
    destinations=["Philadelphia, PA", "Boston, MA"],
    mode="driving"
)

for row in matrix['rows']:
    for element in row['elements']:
        print(f"Distance: {element['distance']['text']}, Duration: {element['duration']['text']}")
````

### 4. Other functions (Optional)

#### 4.1. Places API (Find nearby places)

````python
places_result = gmaps.places_nearby(
    location=(37.4221, -122.0841),
    radius=1000,
    type='restaurant'
)

for place in places_result['results']:
    print(place['name'], "-", place.get('vicinity', 'N/A'))
````

#### 4.2. Autocomplete and Place Details

````python
# Autocomplete example

autocomplete_result = gmaps.places_autocomplete("Starb", types=["establishment"])

print([r['description'] for r in autocomplete_result])
````

````python
# Get place details

place_id = autocomplete_result[0]['place_id']
details = gmaps.place(place_id=place_id)

print(details['result']['name'], details['result']['formatted_address'])
````
