### 1. Using OpenAI API

For details, see offical documents at https://github.com/openai/openai-python.

````python
import os
from openai import OpenAI
from dotenv import load_dotenv
load_dotenv()
````

The best practice to handle API keys and passwords is to keep them out of your main Python scripts. The most common approach is to create a `.env` file which contains:
```
OPENAI_API_KEY=sk-xxxxxxxxxxxxxxxx
Other_Passwards=xxxxxxxxxxxxxx
```

````python
client = OpenAI(api_key = os.getenv("OPENAI_API_KEY"))
````

#### 1.1. Basic Text Prompt

````python
response = client.responses.create(
    model="gpt-4o",
    input="Plan a one-day trip to Sydney, Australia",
)
print(response.output_text)
````

````python
response = client.responses.create(
    model="gpt-4o",
    instructions="You are a teaching assistant for MBA program.",
    input="Summarize the findings of Sloan (1996) 'Do Stock Prices Fully Reflect Information in Accruals and Cash Flows About Future Earnings?' in one paragraph.",
)
print(response.output_text)
````

````python
list_of_CEOs = ['Elon Musk', "Mark Zeckerburg"]
birthday_list = []
for ceo in list_of_CEOs:
    response = client.responses.create(
    model="gpt-4o",
    input=f"What's the birthday of {ceo}, just provide a date (YYYYMMDD).",
    )
    birthday_list.append(response.output_text)
````

#### 1.2. Structured JSON output

````python
schema_hint = """Return JSON with keys:
- "term": string, 
- "definition": string,
- "confidence": float in [0,1]
"""

resp = client.responses.create(
    model="gpt-4o",
    instructions = "You must reply with strict JSON only.",
    input = f"{schema_hint}\nDefine accruals in accounting.",
    temperature=0,
)

text = resp.output_text
# then json.loads(text) in your code; add try/except for safety
print(text)
````

````python
text= text.strip("```").strip('json').strip()
print(text)
````

````python
import json
output = json.loads(text)
output
````

### 2. Using Google Gemini API

- Get API Key from Google AI Studio: 

    1. Visit [Google AI Studio](https://aistudio.google.com/api-keys) and sign in with a Google account; 

    2. Create a project & get API key: click "Get API Key" (under API Access); save your API key in `.env` file.

- Install the official google-generativeai package:

```bash
pip install google-generativeai
```

````python
import google.generativeai as genai
````

````python
genai.configure(api_key = os.getenv("GOOGLE_GAI_API"))

model = genai.GenerativeModel('gemini-2.5-flash')
````

````python
# List available models and capabilities:

for m in genai.list_models():
    print(f"{m.name} - {m.supported_generation_methods}")
````

#### 2.1. Basic Text Prompt

````python
response = model.generate_content("Explain quantum computing in simple terms.")
print(response.text)
````

````python
# For Stateful Conversations:

chat = model.start_chat()

response = chat.send_message("What's the capital of France?")
print(response.text)
````

````python
response = chat.send_message("What's its population?")
print(response.text)
````

#### 2.2 Multimodal Input (e.g., Images)

````python
from PIL import Image

img = Image.open("path_to_image.jpg")

response = model.generate_content([img, "Describe what you see."])
print(response.text)
````
