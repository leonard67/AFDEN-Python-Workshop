**Learning Outcomes**

- Core OOP concepts: **attributes** and **methods**, **encapsulation**
- How to understand various python instances from the perspecitve of OOP

### 1. Define a Simple Class

```python
class FirmYear:
    def __init__(self, gvkey, year):    # Constructor
        self.gvkey = gvkey              # Attributes
        self.year = year

    def info(self):                     # Method
        print(f"This is the observation for gvkey {self.gvkey} in fiscal year {self.year}.")
```

### 2. Create an Object (Instance)

```python
f1 = FirmYear(1004, 2024)
```

```python
# Attribute
print(f1.gvkey)
print(f1.year)

# Method
f1.info()
```
